```python
# Filter for country codes only (leave out regional data)
# Return unique location keys for dataset 1
def get_unique_location_keys(data1):
    return list(set(data1["location_key"].dropna().str.split("_").str[0]))

unique_Country_Keys1 = get_unique_location_keys(data1)

print(unique_Country_Keys1)
print(len(unique_Country_Keys1))
```

```python
# Filter for country codes only (leave out regional data)
# Return unique location keys for dataset 2
def get_unique_location_keys(data2):
    return list(set(data2["location_key"].dropna().str.split("_").str[0]))

unique_Country_Keys2 = get_unique_location_keys(data2)

print(unique_Country_Keys2)
print(len(unique_Country_Keys2))
```

```python
# Function to convert location_key from ISO-2 to ISO-3 format
def convert_iso2_to_iso3(location_key):
    """
    Convert a location_key from ISO-2 to ISO-3 format.
    """
    if pd.isna(location_key):
        return location_key
    
    parts = location_key.split("_")
    iso2 = parts[0]
    
    country = pycountry.countries.get(alpha_2=iso2)
    
    if country:
        parts[0] = country.alpha_3  # Replace ISO-2 with ISO-3
        return "_".join(parts)
    else:
        return location_key  # Return original if not found (e.g. 'XK' for Kosovo)
```

```python
# Pulling out focus data from dataset 1 
# Filtering dataset to only include data by country
focus_data1 = data1[data1['location_key'].isin(unique_Country_Keys1)]
focus_data1.info()
```

```python
# Pulling out focus data from dataset 2
# Filtering dataset to only include data by country
focus_data2 = data2[data2['location_key'].isin(unique_Country_Keys2)]
focus_data2.info()

# Pulling out focus data from dataset 3 (vaccination data)
# Filter for country codes only (leave out regional data)
unique_Country_Keys3 = list(set(data3['location_key'].dropna().str.split('_').str[0]))
focus_data3 = data3[data3['location_key'].isin(unique_Country_Keys3)]
focus_data3.info()
```

```python
# Number of days recorded for each country in dataset 1
focus_data1_numberOfDays=focus_data1['location_key'].value_counts()
focus_data1_numberOfDays
```

```python
# Number of days recorded for each country in datset 2
focus_data2_numberOfDays=focus_data2['location_key'].value_counts()
focus_data2_numberOfDays
```

```python
#Analyzing the distribution of the number of days recorded for each country in both datasets using boxplot statistics

def boxplot_stats(series):
    q1 = series.quantile(0.25)
    median = series.quantile(0.5)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = series[(series < lower_bound) | (series > upper_bound)]

    return {
        "min": series.min(),
        "Q1": q1,
        "median": median,
        "Q3": q3,
        "max": series.max(),
        "IQR": iqr,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "num_outliers": len(outliers)
    }

# Apply to both datasets
stats1 = boxplot_stats(focus_data1_numberOfDays)
stats2 = boxplot_stats(focus_data2_numberOfDays)

# Convert to DataFrame for clean display
summary = pd.DataFrame([stats1, stats2], 
                       index=["Dataset 1", "Dataset 2"])

print(summary)
```

```python
# Using statistics to filter the data
# We will keep countries that have number of days recorded >= Q1 (first quartile)

# For dataset 1
q1_dataset1 = stats1['Q1']
# Filtering dataset to exclude countries with observation numbers below Q1
filtered_countries_1 = focus_data1_numberOfDays[focus_data1_numberOfDays >= q1_dataset1].index.tolist()
filtered_focus_data1 = focus_data1[focus_data1['location_key'].isin(filtered_countries_1)]

print(f"Dataset 1 - Q1 threshold: {q1_dataset1}")
print(f"Countries kept (>= Q1): {len(filtered_countries_1)}")
print(f"Filtered data1 shape: {filtered_focus_data1.shape}")
```

### - Checking for potential missing data

```python
# Check that each country has at least one entity value for every data column in filtered_focus_data1_common.
# This helps verify that each country doesn't have systematic missingness.

def check_country_column_non_missing(df, country_col='location_key', threshold=1):
    """Return non-missing counts, missing-category flags, and a summary per country."""
    # Count non-missing entries per country and column
    non_missing_counts = df.groupby(country_col).apply(lambda group: group.notna().sum())
    if country_col in non_missing_counts.columns:
        non_missing_counts = non_missing_counts.drop(columns=[country_col])

    # Identify columns where the count of non-missing values is below the threshold
    missing_flags = (non_missing_counts < threshold)
    missing_matrix = missing_flags.astype(object)
    missing_matrix[missing_flags] = 'Non'
    missing_matrix[~missing_flags] = ''

    # Build a summary table and list missing categories for each country
    missing_categories = {
        country: non_missing_counts.columns[missing_flags.loc[country]].tolist()
        for country in non_missing_counts.index
    }

    summary = pd.DataFrame({
        'has_minimum_per_column': [len(categories) == 0 for categories in missing_categories.values()],
        'missing_columns': [', '.join(categories) if categories else '' for categories in missing_categories.values()]
    }, index=non_missing_counts.index)

    return non_missing_counts, missing_flags, missing_matrix, summary, missing_categories

# Run the check on filtered_focus_data1_common
non_missing_counts, missing_flags, missing_matrix, country_summary, missing_categories = check_country_column_non_missing(filtered_focus_data1)

print("\nSummary of missing data per category:")
missing_per_category = (missing_matrix == 'Non').sum()
print(missing_per_category)

if country_summary['has_minimum_per_column'].all():
    print("\nAll countries have at least one non-missing value in every column.")
else:
    print("\nCountries with missing entities in one or more categories:")
    for country, categories in missing_categories.items():
        if categories:
            print(f"- {country}: missing data in {', '.join(categories)}")
```

```python
# Function to remove 'new_recovered' and 'cumulative_recovered' columns from the dataframe
def remove_recovered_columns(df):
    """
    Remove 'new_recovered' and 'cumulative_recovered' columns if they exist in the dataframe.
    """
    columns_to_remove = ['new_recovered', 'cumulative_recovered']
    df = df.drop(columns=[col for col in columns_to_remove if col in df.columns], errors='ignore')
    return df

# Apply the function to filtered_focus_data1_common
filtered_focus_data1_common = remove_recovered_columns(filtered_focus_data1)

print(f"After removing columns with too few datapoints, shape: {filtered_focus_data1_common.shape}")
print("Columns remaining:", list(filtered_focus_data1_common.columns))
```

### - Streamlining data to include same dates

```python
# For dataset 1
# Finding the first common observation date shared for all of the countries to make the comparison fair

filtered_focus_data1 = filtered_focus_data1.copy()

filtered_focus_data1['date'] = pd.to_datetime(filtered_focus_data1['date'].astype('string'), errors='coerce').dt.normalize()

min_dates_1 = filtered_focus_data1.groupby('location_key')['date'].min()
max_dates_1 = filtered_focus_data1.groupby('location_key')['date'].max()

first_common_1 = min_dates_1.max()
last_common_1 = max_dates_1.min()

print(f"First common date: {first_common_1}")
print(f"Last common date: {last_common_1}")

# Filtering the filtered_focus_data1 to only include this data interval
filtered_focus_data1_common = filtered_focus_data1[
    (filtered_focus_data1['date'] >= first_common_1) &
    (filtered_focus_data1['date'] <= last_common_1)
].copy()

print(f"filtered_focus_data1_common shape: {filtered_focus_data1_common.shape}")

# Investigate if some dates are missing for some countries. Print the dates and countries missing for the date interval

date_range = pd.date_range(start=first_common_1, end=last_common_1, freq='D')

missing_by_country_1 = {}
for country, group in filtered_focus_data1_common.groupby('location_key'):
    missing_dates = date_range.difference(pd.DatetimeIndex(group['date']))
    if len(missing_dates):
        missing_by_country_1[country] = missing_dates

if missing_by_country_1:
    print("\nMissing dates in filtered_focus_data1_common:")
    for country, missing_dates in missing_by_country_1.items():
        print(f"{country}: {len(missing_dates)} missing dates (first few: {missing_dates[:5].tolist()})")
else:
    print("\nNo missing dates found in filtered_focus_data1_common.")
```

### - Saving the cleaned dataset

```python
#Saving this filterede data for data 1 as a CSV file
output_path = current_dir.parent / 'data' / 'epidemiologi_global_covid_data.csv'
filtered_focus_data1_common.to_csv(output_path, index=False)
print(f"\nSaved merged dataset to: {output_path}")

filtered_focus_data1_common.head()
```

### - Adding vaccination data

```python
# Merge vaccination data into the filtered merged dataset

def merge_vaccination_data(main_df, vaccination_df):
    """
    Merge vaccination data into a COVID analysis dataframe
    on date and location_key.
    """
    main_df = main_df.copy()
    vax_data = vaccination_df.copy()

    main_df['date'] = pd.to_datetime(
        main_df['date'],
        errors='coerce'
    ).dt.normalize()

    vax_data['date'] = pd.to_datetime(
        vax_data['date'],
        errors='coerce'
    ).dt.normalize()

    merged_with_vax = main_df.merge(
    vax_data,
    on=['date', 'location_key'],
    how='left',
    validate='many_to_one'
    )

    return merged_with_vax
```
