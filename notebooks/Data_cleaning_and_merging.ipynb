```python
#Analyzing the distribution of the number of days recorded for each country in both datasets using boxplot statistics

def boxplot_stats(series):
    q1 = series.quantile(0.25)
    median = series.quantile(0.5)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = series[(series < lower_bound) | (series > upper_bound)]

    return {
        "min": series.min(),
        "Q1": q1,
        "median": median,
        "Q3": q3,
        "max": series.max(),
        "IQR": iqr,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "num_outliers": len(outliers)
    }

# Apply to both datasets
stats1 = boxplot_stats(focus_data1_numberOfDays)
stats2 = boxplot_stats(focus_data2_numberOfDays)

# Convert to DataFrame for clean display
summary = pd.DataFrame([stats1, stats2], 
                       index=["Dataset 1", "Dataset 2"])

print(summary)
```

```python
# Using statistics to filter the data
# We will keep countries that have number of days recorded >= Q1 (first quartile)

# For dataset 1
q1_dataset1 = stats1['Q1']
# Filtering dataset to exclude data with observation numbers below Q1
filtered_countries_1 = focus_data1_numberOfDays[focus_data1_numberOfDays >= q1_dataset1].index.tolist()
filtered_focus_data1 = focus_data1[focus_data1['location_key'].isin(filtered_countries_1)]

print(f"Dataset 1 - Q1 threshold: {q1_dataset1}")
print(f"Countries kept (>= Q1): {len(filtered_countries_1)}")
print(f"Filtered data1 shape: {filtered_focus_data1.shape}")

# For dataset 2
q1_dataset2 = stats2['Q1']
# Filtering dataset to exclude data with observation numbers below Q1
filtered_countries_2 = focus_data2_numberOfDays[focus_data2_numberOfDays >= q1_dataset2].index.tolist()
filtered_focus_data2 = focus_data2[focus_data2['location_key'].isin(filtered_countries_2)]

print(f"\nDataset 2 - Q1 threshold: {q1_dataset2}")
print(f"Countries kept (>= Q1): {len(filtered_countries_2)}")
print(f"Filtered data2 shape: {filtered_focus_data2.shape}")
```

```python
#Converting ISO2 country codes to ISO3 for later data visualizations

special_cases = {
    "XK": "XKX",  # Kosovo
}

def iso2_to_iso3(code):
    if pd.isna(code):
        return None
    code = str(code).strip().upper()
    if code in special_cases:
        return special_cases[code]
    country = pycountry.countries.get(alpha_2=code)
    return country.alpha_3 if country else None

#Verifying contrycode conversion
latest_cases_check = (
    merged_data.sort_values("date")
    .groupby("location_key", as_index=False)["cumulative_confirmed"]
    .last()
    .copy()
)

latest_cases_check["iso3"] = latest_cases_check["location_key"].apply(iso2_to_iso3)

print("Total country rows:", len(latest_cases_check))
print("Mapped to ISO-3:", latest_cases_check["iso3"].notna().sum())
print("Failed mappings:", latest_cases_check["iso3"].isna().sum())

failed_codes = sorted(latest_cases_check.loc[latest_cases_check["iso3"].isna(), "location_key"].unique())
print("Failed codes:", failed_codes)
```
